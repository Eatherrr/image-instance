```python
import os, glob, time, yaml, shutil
import cv2
import torch
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from PIL import Image, ExifTags
from torchvision.transforms import functional as F
from torchvision.models.detection import maskrcnn_resnet50_fpn_v2
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor
from torchvision.models.detection.mask_rcnn import MaskRCNNPredictor
from pycocotools.coco import COCO
from pycocotools.cocoeval import COCOeval
from pycocotools import mask as mask_utils
from ultralytics import YOLO

CONF_THRESH = 0.25
WORK_DIR = "/kaggle/working/hasil_evaluasi_cpmk3"
shutil.rmtree(WORK_DIR, ignore_errors=True)
os.makedirs(WORK_DIR, exist_ok=True)
```

```python
# ==========================================================
# 1. DETEKSI DATASET (VALID ASLI + TEST KOS) & BOBOT MODEL
# ==========================================================
yaml_files = glob.glob("/kaggle/input/**/yolooo8/**/data.yaml", recursive=True)
coco_valid_files = glob.glob("/kaggle/input/**/coco-segmentation/**/valid/_annotations.coco.json", recursive=True)
coco_test_files = glob.glob("/kaggle/input/**/coco-segmentation/**/test/_annotations.coco.json", recursive=True)
yolo_base_dirs = glob.glob("/kaggle/input/**/runs/segment/Evaluasi_CPMK3", recursive=True)
rcnn_base_dirs = glob.glob("/kaggle/input/**/kaggle/working/Evaluasi_CPMK3", recursive=True)

assert yaml_files and coco_valid_files and coco_test_files, "Dataset yolooo8 atau coco-segmentation belum lengkap!"
assert yolo_base_dirs and rcnn_base_dirs, "Folder bobot Evaluasi_CPMK3 tidak ditemukan!"

yolo_root = os.path.dirname(yaml_files[0])
coco_orig_json = coco_valid_files[0]
coco_orig_img_dir = os.path.dirname(coco_orig_json)
coco_kos_json = coco_test_files[0]
coco_kos_img_dir = os.path.dirname(coco_kos_json)

yolo_exp_root = yolo_base_dirs[0]
rcnn_exp_root = rcnn_base_dirs[0]

with open(yaml_files[0], "r") as f:
    base_yaml_cfg = yaml.safe_load(f)

def buat_split_yolo(nama_split, daftar_pasangan_img_lbl):
    split_dir = f"/kaggle/working/{nama_split}"
    shutil.rmtree(split_dir, ignore_errors=True)
    os.makedirs(f"{split_dir}/images", exist_ok=True)
    os.makedirs(f"{split_dir}/labels", exist_ok=True)
    for idx, (img_p, lbl_p) in enumerate(daftar_pasangan_img_lbl):
        fname = f"{idx:04d}_{os.path.basename(img_p)}"
        stem = os.path.splitext(fname)[0]
        os.symlink(img_p, os.path.join(f"{split_dir}/images", fname))
        if os.path.exists(lbl_p):
            os.symlink(lbl_p, os.path.join(f"{split_dir}/labels", f"{stem}.txt"))
    cfg = dict(base_yaml_cfg)
    cfg["path"] = split_dir
    cfg["train"] = "images"
    cfg["val"] = "images"
    cfg["test"] = "images"
    yaml_out = f"/kaggle/working/{nama_split}.yaml"
    with open(yaml_out, "w") as f:
        yaml.safe_dump(cfg, f)
    return yaml_out

orig_img_yolo = sorted(glob.glob(f"{yolo_root}/valid/images/*.*"))
kos_img_yolo = sorted(glob.glob(f"{yolo_root}/test/images/*.*"))

pasangan_orig = [
    (p, os.path.join(yolo_root, "valid/labels", f"{os.path.splitext(os.path.basename(p))[0]}.txt"))
    for p in orig_img_yolo
]
pasangan_kos = [
    (p, os.path.join(yolo_root, "test/labels", f"{os.path.splitext(os.path.basename(p))[0]}.txt"))
    for p in kos_img_yolo
]

yaml_orig = buat_split_yolo("split_yolo_orig", pasangan_orig)
yaml_kos = buat_split_yolo("split_yolo_kos40", pasangan_kos)
yaml_all = buat_split_yolo("split_yolo_all", pasangan_orig + pasangan_kos)

print(f"YOLO -> Valid Asli: {len(orig_img_yolo)} | Test Kos: {len(kos_img_yolo)} | Total: {len(orig_img_yolo)+len(kos_img_yolo)}")
```

```python
# ==========================================================
# 2. EKSTRAKSI EXIF / METADATA 40 FOTO KOS (BUKTI KOMPONEN 1)
# ==========================================================
exif_rows = []
for p in kos_img_yolo:
    with Image.open(p) as im:
        w, h = im.size
        raw_exif = im.getexif()
        row_ex = {"file_name": os.path.basename(p), "width": w, "height": h, "format": im.format}
        if raw_exif:
            for tag_id, val in raw_exif.items():
                tag_name = ExifTags.TAGS.get(tag_id, str(tag_id))
                if tag_name in ("Make", "Model", "DateTimeOriginal", "ExposureTime", "FNumber", "ISOSpeedRatings", "FocalLength"):
                    row_ex[tag_name] = str(val)
        exif_rows.append(row_ex)
df_exif = pd.DataFrame(exif_rows)
df_exif.to_csv(f"{WORK_DIR}/exif_metadata_40_foto.csv", index=False)

rekap_utama = []
ap50_per_kelas = {}
ap95_per_kelas = {}
loaded_yolo_models = {}
loaded_rcnn_models = {}
```

```python
# ==========================================================
# 3. EVALUASI 6 MODEL YOLO (YOLOv8 & YOLO11)
# ==========================================================
daftar_yolo = [
    ("YOLOv8", "Tanpa Aug", "YOLOv8_Tanpa_Aug"),
    ("YOLOv8", "Default Aug", "YOLOv8_Default_Aug"),
    ("YOLOv8", "Custom Aug", "YOLOv8_Custom_Aug"),
    ("YOLO11", "Tanpa Aug", "YOLO11_Tanpa_Aug"),
    ("YOLO11", "Default Aug", "YOLO11_Default_Aug"),
    ("YOLO11", "Custom Aug", "YOLO11_Custom_Aug"),
]

for ars, aug_lbl, folder_name in daftar_yolo:
    label_model = f"{ars} - {aug_lbl}"
    kandidat = glob.glob(f"{yolo_exp_root}/{folder_name}/**/best.pt", recursive=True)
    if not kandidat:
        print(f"[SKIP] best.pt tidak ditemukan untuk {folder_name}")
        continue

    print(f"\n[YOLO] Mengevaluasi {label_model}...")
    model = YOLO(kandidat[0])
    loaded_yolo_models[folder_name] = model

    m_orig = model.val(data=yaml_orig, split="test", imgsz=640, verbose=False)
    m_kos = model.val(data=yaml_kos, split="test", imgsz=640, verbose=False)
    m_all = model.val(data=yaml_all, split="test", imgsz=640, verbose=False)

    # Ukur latency inferensi murni (setelah 3x warm-up)
    for wp in kos_img_yolo[:3]:
        model.predict(source=wp, imgsz=640, conf=CONF_THRESH, verbose=False)
    t0 = time.perf_counter()
    model.predict(
        source=kos_img_yolo,
        imgsz=640,
        conf=CONF_THRESH,
        save=True,
        project=f"{WORK_DIR}/per_model",
        name=folder_name,
        verbose=False
    )
    dt_ms = ((time.perf_counter() - t0) / max(len(kos_img_yolo), 1)) * 1000.0

    orig_m50, kos_m50 = float(m_orig.seg.map50), float(m_kos.seg.map50)
    orig_m95, kos_m95 = float(m_orig.seg.map), float(m_kos.seg.map)
    drop_m50 = ((orig_m50 - kos_m50) / orig_m50 * 100.0) if orig_m50 > 0 else 0.0
    drop_m95 = ((orig_m95 - kos_m95) / orig_m95 * 100.0) if orig_m95 > 0 else 0.0

    rekap_utama.append({
        "Arsitektur": ars,
        "Augmentasi": aug_lbl,
        "Model": label_model,
        "Orig Mask mAP50": round(orig_m50, 4),
        "Kos40 Mask mAP50": round(kos_m50, 4),
        "Drop mAP50 (%)": round(drop_m50, 1),
        "Orig Mask mAP50-95": round(orig_m95, 4),
        "Kos40 Mask mAP50-95": round(kos_m95, 4),
        "Drop mAP50-95 (%)": round(drop_m95, 1),
        "All Mask mAP50": round(float(m_all.seg.map50), 4),
        "All Box mAP50": round(float(m_all.box.map50), 4),
        "Kos40 Precision": round(float(m_kos.seg.mp), 4),
        "Kos40 Recall": round(float(m_kos.seg.mr), 4),
        "Latency (ms/img)": round(dt_ms, 2),
        "FPS": round(1000.0 / dt_ms, 1) if dt_ms > 0 else 0.0,
    })

    c50, c95 = {}, {}
    for idx_cls, v50, v95 in zip(m_kos.seg.ap_class_index, m_kos.seg.ap50, m_kos.seg.ap):
        cname = model.names[int(idx_cls)]
        c50[cname] = round(float(v50), 4)
        c95[cname] = round(float(v95), 4)
    ap50_per_kelas[label_model] = c50
    ap95_per_kelas[label_model] = c95
```

```python
# ==========================================================
# 4. EVALUASI 3 MODEL MASK R-CNN (ANTI-OOM + PROGRESS BAR)
# ==========================================================
import gc
from tqdm.auto import tqdm

rekap_utama = [r for r in rekap_utama if r["Arsitektur"] != "Mask R-CNN"]
for m_yolo in loaded_yolo_models.values():
    try:
        if hasattr(m_yolo, "model") and m_yolo.model is not None:
            m_yolo.model.cpu()
    except Exception:
        pass
loaded_rcnn_models.clear()
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
coco_gt_orig = COCO(coco_orig_json)
coco_gt_kos = COCO(coco_kos_json)

cat_ids_train = sorted(coco_gt_orig.getCatIds())
id_to_name = {c["id"]: c["name"] for c in coco_gt_orig.loadCats(cat_ids_train)}
name_to_kos_cid = {c["name"]: c["id"] for c in coco_gt_kos.loadCats(coco_gt_kos.getCatIds())}

orig_img_ids = coco_gt_orig.getImgIds()
kos_img_ids = coco_gt_kos.getImgIds()
print(f"COCO -> Valid Asli: {len(orig_img_ids)} | Test Kos: {len(kos_img_ids)}")

def jalankan_inferensi_coco(model, coco_obj, img_dir, is_kos=False, vis_dir=None, simpan_count=False, max_side=1280):
    results = []
    latencies = []
    desc = "Foto Kos (40)" if is_kos else "Valid Asli (230)"
    with torch.no_grad():
        for idx_i, img_id in enumerate(tqdm(coco_obj.getImgIds(), desc=desc)):
            info = coco_obj.loadImgs(img_id)[0]
            img_path = os.path.join(img_dir, info["file_name"])
            pil_img = Image.open(img_path).convert("RGB")
            orig_w, orig_h = pil_img.size

            scale = min(1.0, max_side / max(orig_w, orig_h))
            if scale < 1.0:
                new_w, new_h = int(round(orig_w * scale)), int(round(orig_h * scale))
                infer_img = pil_img.resize((new_w, new_h), Image.BILINEAR)
            else:
                new_w, new_h = orig_w, orig_h
                infer_img = pil_img

            tensor_img = F.to_tensor(infer_img).to(device)

            if device.type == "cuda":
                torch.cuda.synchronize()
            t0 = time.perf_counter()
            out = model([tensor_img])[0]
            if device.type == "cuda":
                torch.cuda.synchronize()
            dt = (time.perf_counter() - t0) * 1000.0

            if is_kos and idx_i >= 3:
                latencies.append(dt)

            boxes = out["boxes"].detach().cpu().numpy()
            labels = out["labels"].detach().cpu().numpy()
            scores = out["scores"].detach().cpu().numpy()
            masks = out["masks"].detach().cpu().numpy()
            del out, tensor_img

            if is_kos and vis_dir is not None:
                vis_img = cv2.cvtColor(np.array(infer_img), cv2.COLOR_RGB2BGR).copy()
                det_valid = 0
                for box, lbl, scr, msk in zip(boxes, labels, scores, masks):
                    if scr < CONF_THRESH or not (1 <= int(lbl) <= len(cat_ids_train)):
                        continue
                    det_valid += 1
                    x1, y1, x2, y2 = map(int, box)
                    cid = cat_ids_train[int(lbl) - 1]
                    cls_name = id_to_name.get(cid, str(cid))
                    color = np.random.RandomState(int(lbl) * 17).randint(60, 240, size=3).tolist()

                    bin_m = (msk[0] > 0.5).astype(np.uint8)
                    vis_img[bin_m == 1] = vis_img[bin_m == 1] * 0.5 + np.array(color) * 0.5
                    cv2.rectangle(vis_img, (x1, y1), (x2, y2), color, 2)

                    txt = f"{cls_name} {scr:.2f}"
                    (tw, th), _ = cv2.getTextSize(txt, cv2.FONT_HERSHEY_SIMPLEX, 0.55, 1)
                    cv2.rectangle(vis_img, (x1, max(0, y1 - th - 6)), (x1 + tw + 4, y1), color, -1)
                    cv2.putText(vis_img, txt, (x1 + 2, max(th + 2, y1 - 4)),
                                cv2.FONT_HERSHEY_SIMPLEX, 0.55, (255, 255, 255), 1, cv2.LINE_AA)
                cv2.imwrite(os.path.join(vis_dir, info["file_name"]), vis_img)
                if simpan_count:
                    rcnn_det_counts_kos[img_id] = det_valid

            scale_x = orig_w / new_w
            scale_y = orig_h / new_h
            for box, lbl, scr, msk in zip(boxes, labels, scores, masks):
                if scr < 0.05 or not (1 <= int(lbl) <= len(cat_ids_train)):
                    continue
                x1, y1, x2, y2 = box.tolist()
                x1, x2 = x1 * scale_x, x2 * scale_x
                y1, y2 = y1 * scale_y, y2 * scale_y

                cid = cat_ids_train[int(lbl) - 1]
                cls_name = id_to_name[cid]
                target_cid = name_to_kos_cid.get(cls_name, cid) if is_kos else cid

                bin_mask = (msk[0] > 0.5).astype(np.uint8)
                if scale < 1.0:
                    bin_mask = cv2.resize(bin_mask, (orig_w, orig_h), interpolation=cv2.INTER_NEAREST)
                rle = mask_utils.encode(np.asfortranarray(bin_mask))
                rle["counts"] = rle["counts"].decode("utf-8")
                results.append({
                    "image_id": img_id,
                    "category_id": target_cid,
                    "bbox": [x1, y1, x2 - x1, y2 - y1],
                    "score": float(scr),
                    "segmentation": rle
                })

            del boxes, labels, scores, masks
            if device.type == "cuda" and (idx_i + 1) % 10 == 0:
                torch.cuda.empty_cache()

    return results, latencies

def hitung_coco_stats(coco_gt_obj, results_list, iou_type):
    if not results_list:
        return 0.0, 0.0, None
    coco_dt_obj = coco_gt_obj.loadRes(results_list)
    ev = COCOeval(coco_gt_obj, coco_dt_obj, iou_type)
    ev.evaluate()
    ev.accumulate()
    ev.summarize()
    return float(ev.stats[1]), float(ev.stats[0]), ev

daftar_rcnn = [
    ("Mask R-CNN", "Tanpa Aug", "MaskRCNN_Tanpa_Aug"),
    ("Mask R-CNN", "Default Aug", "MaskRCNN_Default_Aug"),
    ("Mask R-CNN", "Custom Aug", "MaskRCNN_Custom_Aug"),
]

rcnn_det_counts_kos = {}

for ars, aug_lbl, folder_name in daftar_rcnn:
    label_model = f"{ars} - {aug_lbl}"
    w_path = f"{rcnn_exp_root}/{folder_name}/best.pth"
    if not os.path.exists(w_path):
        print(f"[SKIP] best.pth tidak ditemukan di {w_path}")
        continue

    print(f"\n[Mask R-CNN] Mengevaluasi {label_model}...")
    state_dict = torch.load(w_path, map_location="cpu")
    num_classes = state_dict["roi_heads.box_predictor.cls_score.weight"].shape[0]

    model = maskrcnn_resnet50_fpn_v2(weights=None, weights_backbone=None)
    model.roi_heads.box_predictor = FastRCNNPredictor(
        model.roi_heads.box_predictor.cls_score.in_features, num_classes
    )
    model.roi_heads.mask_predictor = MaskRCNNPredictor(
        model.roi_heads.mask_predictor.conv5_mask.in_channels, 256, num_classes
    )
    model.load_state_dict(state_dict)
    model.to(device).eval()

    vis_dir = f"{WORK_DIR}/per_model/{folder_name}"
    os.makedirs(vis_dir, exist_ok=True)

    res_orig, _ = jalankan_inferensi_coco(model, coco_gt_orig, coco_orig_img_dir, is_kos=False)
    res_kos, kos_latencies = jalankan_inferensi_coco(
        model, coco_gt_kos, coco_kos_img_dir, is_kos=True, vis_dir=vis_dir,
        simpan_count=(folder_name == "MaskRCNN_Custom_Aug")
    )

    orig_m50, orig_m95, _ = hitung_coco_stats(coco_gt_orig, res_orig, "segm")
    kos_m50, kos_m95, ev_kos_seg = hitung_coco_stats(coco_gt_kos, res_kos, "segm")
    orig_b50, _, _ = hitung_coco_stats(coco_gt_orig, res_orig, "bbox")
    kos_b50, _, _ = hitung_coco_stats(coco_gt_kos, res_kos, "bbox")

    n_o, n_k = len(orig_img_ids), len(kos_img_ids)
    all_m50 = (orig_m50 * n_o + kos_m50 * n_k) / max(n_o + n_k, 1)
    all_b50 = (orig_b50 * n_o + kos_b50 * n_k) / max(n_o + n_k, 1)

    drop_m50 = ((orig_m50 - kos_m50) / orig_m50 * 100.0) if orig_m50 > 0 else 0.0
    drop_m95 = ((orig_m95 - kos_m95) / orig_m95 * 100.0) if orig_m95 > 0 else 0.0
    avg_ms = float(np.mean(kos_latencies)) if kos_latencies else 0.0

    mp, mr = 0.0, 0.0
    if ev_kos_seg is not None:
        prec_50 = ev_kos_seg.eval["precision"][0, :, :, 0, 2]
        rec_50 = ev_kos_seg.eval["recall"][0, :, 0, 2]
        mp = float(np.mean(prec_50[prec_50 > -1])) if np.any(prec_50 > -1) else 0.0
        mr = float(np.mean(rec_50[rec_50 > -1])) if np.any(rec_50 > -1) else 0.0

        c50, c95 = {}, {}
        for c_idx, c_id in enumerate(sorted(coco_gt_kos.getCatIds())):
            cname = coco_gt_kos.loadCats([c_id])[0]["name"]
            p50 = ev_kos_seg.eval["precision"][0, :, c_idx, 0, 2]
            p95 = ev_kos_seg.eval["precision"][:, :, c_idx, 0, 2]
            if np.any(p50 > -1):
                c50[cname] = round(float(np.mean(p50[p50 > -1])), 4)
                c95[cname] = round(float(np.mean(p95[p95 > -1])), 4)
        ap50_per_kelas[label_model] = c50
        ap95_per_kelas[label_model] = c95

    rekap_utama.append({
        "Arsitektur": ars,
        "Augmentasi": aug_lbl,
        "Model": label_model,
        "Orig Mask mAP50": round(orig_m50, 4),
        "Kos40 Mask mAP50": round(kos_m50, 4),
        "Drop mAP50 (%)": round(drop_m50, 1),
        "Orig Mask mAP50-95": round(orig_m95, 4),
        "Kos40 Mask mAP50-95": round(kos_m95, 4),
        "Drop mAP50-95 (%)": round(drop_m95, 1),
        "All Mask mAP50": round(all_m50, 4),
        "All Box mAP50": round(all_b50, 4),
        "Kos40 Precision": round(mp, 4),
        "Kos40 Recall": round(mr, 4),
        "Latency (ms/img)": round(avg_ms, 2),
        "FPS": round(1000.0 / avg_ms, 1) if avg_ms > 0 else 0.0,
    })

    model.cpu()
    del model, state_dict, res_orig, res_kos
    gc.collect()
    if device.type == "cuda":
        torch.cuda.empty_cache()

df_utama = pd.DataFrame(rekap_utama)
df_utama.to_csv(f"{WORK_DIR}/rekap_evaluasi_9_model.csv", index=False)
pd.DataFrame(ap50_per_kelas).fillna(0.0).to_csv(f"{WORK_DIR}/rekap_ap50_per_kelas_kos40.csv")
pd.DataFrame(ap95_per_kelas).fillna(0.0).to_csv(f"{WORK_DIR}/rekap_ap50_95_per_kelas_kos40.csv")
print("\nEvaluasi 3 model Mask R-CNN selesai!")
```

```python
import os, glob, cv2, torch
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import torchvision.transforms.functional as F
from torchvision.models.detection import maskrcnn_resnet50_fpn_v2
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor
from torchvision.models.detection.mask_rcnn import MaskRCNNPredictor
from pycocotools.coco import COCO
from ultralytics import YOLO
import ipywidgets as widgets
from IPython.display import display

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 1. Load Mapping Kelas COCO (buat label Mask R-CNN)
coco_gt_orig = COCO(coco_orig_json)
cat_ids_train = sorted(coco_gt_orig.getCatIds())
id_to_name = {c["id"]: c["name"] for c in coco_gt_orig.loadCats(cat_ids_train)}

# 2. Load YOLOv8 & YOLO11 Custom Aug
path_y8  = glob.glob(f"{yolo_exp_root}/YOLOv8_Custom_Aug/**/best.pt", recursive=True)[0]
path_y11 = glob.glob(f"{yolo_exp_root}/YOLO11_Custom_Aug/**/best.pt", recursive=True)[0]
model_y8  = YOLO(path_y8)
model_y11 = YOLO(path_y11)

# 3. Load Mask R-CNN Custom Aug
w_rcnn = f"{rcnn_exp_root}/MaskRCNN_Custom_Aug/best.pth"
state_dict = torch.load(w_rcnn, map_location="cpu")
num_classes = state_dict["roi_heads.box_predictor.cls_score.weight"].shape[0]

model_rcnn = maskrcnn_resnet50_fpn_v2(weights=None, weights_backbone=None)
model_rcnn.roi_heads.box_predictor = FastRCNNPredictor(
    model_rcnn.roi_heads.box_predictor.cls_score.in_features, num_classes
)
model_rcnn.roi_heads.mask_predictor = MaskRCNNPredictor(
    model_rcnn.roi_heads.mask_predictor.conv5_mask.in_channels, 256, num_classes
)
model_rcnn.load_state_dict(state_dict)
model_rcnn.to(device).eval()

print("3 Model (YOLOv8, YOLO11, Mask R-CNN) SIAP DEMO!")

# 4. Fungsi Live Demo 4 Panel (Samping-Sampingan)
def live_demo(img_path, conf=0.25, imgsz=640, max_side=1280):
    # --- YOLO ---
    res8  = model_y8.predict(source=img_path, conf=conf, imgsz=imgsz, verbose=False)[0]
    res11 = model_y11.predict(source=img_path, conf=conf, imgsz=imgsz, verbose=False)[0]
    vis8  = cv2.cvtColor(res8.plot(), cv2.COLOR_BGR2RGB)
    vis11 = cv2.cvtColor(res11.plot(), cv2.COLOR_BGR2RGB)

    # --- Mask R-CNN (Anti-OOM Resize) ---
    pil_img = Image.open(img_path).convert("RGB")
    orig_w, orig_h = pil_img.size
    scale = min(1.0, max_side / max(orig_w, orig_h))
    infer_img = pil_img.resize((int(round(orig_w * scale)), int(round(orig_h * scale))), Image.BILINEAR) if scale < 1.0 else pil_img

    tensor_img = F.to_tensor(infer_img).to(device)
    with torch.no_grad():
        out = model_rcnn([tensor_img])[0]

    boxes  = out["boxes"].detach().cpu().numpy()
    labels = out["labels"].detach().cpu().numpy()
    scores = out["scores"].detach().cpu().numpy()
    masks  = out["masks"].detach().cpu().numpy()

    vis_rcnn = np.array(infer_img).copy()
    det_rcnn = 0
    for box, lbl, scr, msk in zip(boxes, labels, scores, masks):
        if scr < conf or not (1 <= int(lbl) <= len(cat_ids_train)):
            continue
        det_rcnn += 1
        x1, y1, x2, y2 = map(int, box)
        cid = cat_ids_train[int(lbl) - 1]
        cls_name = id_to_name.get(cid, str(cid))
        color = np.random.RandomState(int(lbl) * 17).randint(60, 240, size=3).tolist()

        bin_m = (msk[0] > 0.5).astype(np.uint8)
        vis_rcnn[bin_m == 1] = (vis_rcnn[bin_m == 1] * 0.5 + np.array(color) * 0.5).astype(np.uint8)
        cv2.rectangle(vis_rcnn, (x1, y1), (x2, y2), color, 2)

        txt = f"{cls_name} {scr:.2f}"
        (tw, th), _ = cv2.getTextSize(txt, cv2.FONT_HERSHEY_SIMPLEX, 0.55, 1)
        cv2.rectangle(vis_rcnn, (x1, max(0, y1 - th - 6)), (x1 + tw + 4, y1), color, -1)
        cv2.putText(vis_rcnn, txt, (x1 + 2, max(th + 2, y1 - 4)),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.55, (255, 255, 255), 1, cv2.LINE_AA)

    # --- Plot 4 Panel Side-by-Side ---
    fig, axes = plt.subplots(1, 4, figsize=(24, 6))
    axes[0].imshow(pil_img);  axes[0].set_title("Foto Asli"); axes[0].axis("off")
    axes[1].imshow(vis8);     axes[1].set_title(f"YOLOv8 Custom (conf={conf}) | Det: {len(res8.boxes)}"); axes[1].axis("off")
    axes[2].imshow(vis11);    axes[2].set_title(f"YOLO11 Custom (conf={conf}) | Det: {len(res11.boxes)}"); axes[2].axis("off")
    axes[3].imshow(vis_rcnn); axes[3].set_title(f"Mask R-CNN Custom (conf={conf}) | Det: {det_rcnn}"); axes[3].axis("off")
    plt.tight_layout()
    plt.show()

# 5. Munculkan Tombol Upload
uploader = widgets.FileUpload(accept='image/*', multiple=False)
display(uploader)
```
